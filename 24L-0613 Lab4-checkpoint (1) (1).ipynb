```python
from queue import Queue, PriorityQueue

def read_cube(filename):
    cube = []
    with open(filename, 'r') as f:
        for line in f:
            cube.append(list(map(int, line.strip().split())))
    return cube

class Node:
    def __init__(self, x, y, parent=None, cost=0):
        self.state = (x, y)
        self.parent = parent
        self.cost = cost

    def getneighbours(self, board):
        neighbours = []
        x, y = self.state

        if x+1 < len(board) and board[x+1][y] == 0:
            neighbours.append(Node(x+1, y, self, self.cost+1))

        if x-1 >= 0 and board[x-1][y] == 0:
            neighbours.append(Node(x-1, y, self, self.cost+1))

        if y+1 < len(board[0]) and board[x][y+1] == 0:
            neighbours.append(Node(x, y+1, self, self.cost+1))

        if y-1 >= 0 and board[x][y-1] == 0:
            neighbours.append(Node(x, y-1, self, self.cost+1))

        return neighbours

def get_path(node):
    path = []
    while node:
        path.append(node.state)
        node = node.parent
    return path[::-1]

def bfs(board):
    start = Node(0, 0)
    goal = (len(board)-1, len(board[0])-1)

    visited = set()
    que = Queue()
    que.put(start)
    visited.add(start.state)

    while not que.empty():
        value = que.get()

        if value.state == goal:
            return get_path(value)

        neighbour = value.getneighbours(board)
        for x in neighbour:
            if x.state not in visited:
                que.put(x)
                visited.add(x.state)

    return -1



def dfs(board):
    start = Node(0, 0)
    goal = (len(board)-1, len(board[0])-1)

    visited = set()
    stack = [start]

    while stack:
        value = stack.pop()

        if value.state == goal:
            return get_path(value)

        if value.state not in visited:
            visited.add(value.state)

        neighbour = value.getneighbours(board)
        for x in neighbour:
            if x.state not in visited:
                stack.append(x)

    return -1

def DLS(start, board, goal, limit):
    stack = [(start, 0)]
    visited = set()

    while stack:
        value, depth = stack.pop()

        if value.state == goal:
            return value

        if value.state not in visited:
            visited.add(value.state)

        if depth < limit:
            neighbour = value.getneighbours(board)
            for x in neighbour:
                if x.state not in visited:
                    stack.append((x, depth+1))

    return None


def ids(board, max_depth=50):
    start = Node(0, 0)
    goal = (len(board)-1, len(board[0])-1)

    for d in range(max_depth):
        res = DLS(start, board, goal, d)
        if res:
            return get_path(res)
    return -1


def ucs(board):
    start = Node(0, 0)
    goal = (len(board)-1, len(board[0])-1)

    visited = set()
    que = PriorityQueue()
    que.put((0, start.state,start))

    while not que.empty():
        cost,_ ,value = que.get()

        if value.state == goal:
            return get_path(value)

        if value.state in visited:
            continue

        visited.add(value.state)

        neighbour = value.getneighbours(board)
        for x in neighbour:
            if x.state not in visited:
                que.put((x.cost,x.state,x))

    return -1

# cube = read_cube("cube.txt")

cube = read_cube("D:\cube.txt")

print("BFS Path:", bfs(cube))
print("DFS Path:", dfs(cube))
print("UCS Path:", ucs(cube))
print("IDS Path:", ids(cube))
```
