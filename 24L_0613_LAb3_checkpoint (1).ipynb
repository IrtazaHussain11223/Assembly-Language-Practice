```python
import numpy as np
arr = np.random.randint(-50, 151, size=100)
arr_reshaped = arr.reshape((10, 10))
arr_reshaped[arr_reshaped < 0] = 0
arr_min = arr_reshaped.min()
arr_max = arr_reshaped.max()
normalized = (arr_reshaped - arr_min) / (arr_max - arr_min)
row_means = normalized.mean(axis=1)
print("Original reshaped array (negatives replaced):\n", arr_reshaped)
print("\nNormalized array:\n", normalized)
print("\nRow-wise means:\n", row_means)
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["Lahore", "Karachi", "Lahore"]
}
df = pd.DataFrame(data)

print(df.head())
print(df.tail())
print(df.shape)
print(df.dtypes)
print(df["Name"])
print(df[df["Name"] == "Bob"])
print(df.iloc[1, 2])
print(df[df["Age"] > 28])

df["Score"] = [85, 90, 95]
print(df)

df.loc[df["Name"] == "Charlie", "City"] = "Multan"
print(df)

df.drop("Score", axis=1, inplace=True)
print(df)

print(df["Age"].mean())
print(df["City"].nunique())
print(df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes.csv")

print(df.shape)
print(df.columns)
print(df.head())
print(df.info())
print(df.describe())

plt.figure()
plt.scatter(df["Glucose"], df["BMI"], c=df["Outcome"])
plt.xlabel("Glucose")
plt.ylabel("BMI")
plt.title("Glucose vs BMI (colored by Outcome)")
plt.show()

plt.figure()
plt.scatter(df["Age"], df["Insulin"], c=df["Outcome"])
plt.xlabel("Age")
plt.ylabel("Insulin")
plt.title("Age vs Insulin (colored by Outcome)")
plt.show()

plt.figure()
plt.scatter(df["BloodPressure"], df["SkinThickness"], c=df["Outcome"])
plt.xlabel("BloodPressure")
plt.ylabel("SkinThickness")
plt.title("BloodPressure vs SkinThickness (colored by Outcome)")
plt.show()

plt.figure()
df.boxplot(column="Glucose", by="Outcome")
plt.show()

plt.figure()
df.boxplot(column="BMI", by="Outcome")
plt.show()

plt.figure()
df.boxplot(column="Insulin", by="Outcome")
plt.show()

plt.figure()
df.boxplot(column="Age", by="Outcome")
plt.show()

df.hist(figsize=(10, 8))
plt.show()

plt.figure(figsize=(8, 6))
plt.imshow(df.corr(), cmap="viridis")
plt.colorbar()
plt.xticks(range(len(df.columns)), df.columns, rotation=90)
plt.yticks(range(len(df.columns)), df.columns)
plt.title("Correlation Matrix")
plt.show()
```
