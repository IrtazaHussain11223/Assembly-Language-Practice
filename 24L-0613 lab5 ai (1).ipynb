```python
from queue import PriorityQueue
def read_cube(filename):
    cube = []
    with open(filename, 'r') as f:
        for line in f:
            cube.append(list(map(int, line.strip().split())))
    return cube
    
def heuristic(x, y, goal):
    return abs(x - goal[0]) + abs(y - goal[1])
# -------- Get Neighbours --------
def get_neighbours(board, x, y, short_walls):
    rows, cols = len(board), len(board[0])
    neighbours = []
    moves = [(1,0), (-1,0), (0,1), (0,-1)]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy

        if 0 <= nx < rows and 0 <= ny < cols:
            if board[nx][ny] == 1:
                continue

            new_short = short_walls
            move_cost = 1
            if board[nx][ny] == 2:
                new_short += 1
                move_cost = 2

            neighbours.append((nx, ny, move_cost, new_short))

    return neighbours
# greedy
def greedy_bfs(board):
    rows, cols = len(board), len(board[0])
    start = (0, 0)
    goal = (rows-1, cols-1)

    pq = PriorityQueue()
    pq.put((heuristic(0, 0, goal), start, [start]))

    visited = set()

    while not pq.empty():
        _, (x, y), path = pq.get()

        if (x, y) == goal:
            return path

        if (x, y) in visited:
            continue

        visited.add((x, y))

        for nx, ny, _, _ in get_neighbours(board, x, y, 0):
            if (nx, ny) not in visited:
                h = heuristic(nx, ny, goal)
                pq.put((h, (nx, ny), path + [(nx, ny)]))

    return -1
#A * search
def a_star(board):
    rows, cols = len(board), len(board[0])
    start = (0, 0)
    goal = (rows-1, cols-1)

    pq = PriorityQueue()
    pq.put((0, 0, start, [start]))
    visited = {}
    while not pq.empty():
        f, short_walls, (x, y), path = pq.get()

        if (x, y) == goal:
            return path

        if (x, y) in visited and visited[(x, y)] <= short_walls:
            continue

        visited[(x, y)] = short_walls

        for nx, ny, move_cost, new_short in get_neighbours(board, x, y, short_walls):

            g = len(path) + move_cost
            h = heuristic(nx, ny, goal)
            new_f = g + h

            pq.put((new_f, new_short, (nx, ny), path + [(nx, ny)]))

    return -1

cube = read_cube("D:\\cube.txt")

print("Greedy BFS Path:", greedy_bfs(cube))
print("A* Path:", a_star(cube))
```
